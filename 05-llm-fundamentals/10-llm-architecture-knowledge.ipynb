## Common Building Blocks

**Definition.** Most LLMs share: tokenizer, embeddings, Pre-Norm Transformer blocks (MHA/GQA + MLP), RoPE-like positions, LM head; variants add MoE, sliding windows, etc.

**Why it matters.** Shared vocabulary lets you transfer mental models across vendors.

**How it works.** Map each block to a serving concern (memory, latency, quality).

**Intuition.** Different brands, same chassis with option packages.

**Common pitfalls.**
- Buzzword shopping without mapping to blocks
- Ignoring tokenizer differences

**When to use.** Architecture reviews and vendor comparisons.

```mermaid
flowchart TB
  Tok[Tokenizer] --> Emb[Embeddings]
  Emb --> Blocks[N x Transformer block]
  Blocks --> Head[LM head]
```

## Reading a Model Card Checklist

**Definition.** A structured pass over license, architecture, context, training data notes, evals, safety, and deploy specs.

**Why it matters.** Prevents shipping surprises on license, length, or languages.

**How it works.** Fill a checklist; escalate unknowns; verify with smoke tests.

**Intuition.** Nutrition label before eating the model.

**Common pitfalls.**
- Skimming only MMLU
- Ignoring license/commercial terms

**When to use.** Every model adoption decision.

### Checklist highlights

| Field | Why |
|-------|-----|
| License | Legal |
| Context | Product UX/cost |
| Cutoff | Staleness |
| Evals | Fit |
| Safety | Risk |
| Quant options | Deploy |
