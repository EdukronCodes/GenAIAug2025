# Local LLM Fundamentals

Running models on your own hardware (or VPC GPUs you control) changes the privacy, cost, latency, and ops calculus versus hosted APIs. This notebook builds the mental model for local/self-hosted inference.

```mermaid
  HW[CPU/GPU/NPU] --> Runtime
```

## 1. Why Local / Self-Hosted?

| Driver | Local win | Hosted win |
|--------|-----------|------------|
| Privacy / residency | Weights & prompts stay in boundary | Vendor BAAs / regions |
| Cost at scale | Amortized GPUs | Pay-per-token simplicity |
| Latency | LAN / same rack | Cold starts / internet |
| Customization | Quantization, LoRA, routers | Fastest features |
| Ops burden | You own uptime | Vendor SLAs |

### Pitfalls
- Underestimating engineering cost of GPU clusters
- Ignoring model license constraints
- Assuming “local” equals “secure” without authn/z


## 2. Inference Mental Model

### Latency vs throughput
- **Latency (interactive):** time-to-first-token (TTFT) + tokens/sec for one user
- **Throughput (batch):** aggregate tokens/sec across concurrent requests

### Prefill vs decode
1. **Prefill** — process prompt (compute-heavy, parallelizable)
2. **Decode** — generate tokens autoregressively (memory-bandwidth heavy)

```
Prompt tokens ──prefill──► KV cache
                              │
                              ▼
                     decode token by token
```


````python
# Demo 1 — Rough VRAM estimate for dense models (rule of thumb)
def estimate_vram_gb(
    params_b: float,
    bits: float = 16,
    kv_overhead_gb: float = 1.5,
    framework_gb: float = 1.0,
) -> float:
    """params_b in billions. 16-bit ~2 bytes/param; 4-bit ~0.5 bytes/param."""
    bytes_per_param = bits / 8.0
    weights_gb = params_b * 1e9 * bytes_per_param / (1024**3)
    return weights_gb + kv_overhead_gb + framework_gb

for p, b in [(7, 16), (7, 4), (70, 4), (70, 16)]:
    print(f"{p}B @ {b}-bit ~ {estimate_vram_gb(p, b):.1f} GB")
````

## 3. Hardware Cheatsheet

| Hardware | Sweet spot | Notes |
|----------|------------|-------|
| Laptop CPU | ≤3–8B Q4 | Slow decode |
| Consumer GPU 8–12GB | 7B Q4–Q5 | Good interactive |
| 24GB GPU | 13–34B Q4 / 7B FP16 | Comfortable |
| 48–80GB | 70B Q4 / multi-LoRA | Team server |
| Multi-GPU | 70B+ FP16 / MoE | Tensor parallel |

### Intuition
Quantization trades precision for fitting in VRAM; quality cliffs are task-dependent—**measure on your evals**.


## 4. Serving Modes

| Mode | Goal | Knobs |
|------|------|-------|
| Interactive chat | Low TTFT | Small batches, KV cache |
| IDE completion | Ultra-low latency | Speculative decode, short ctx |
| Batch offline | Max throughput | Continuous batching |
| Hybrid | Local + cloud failover | Router policies |


````python
# Demo 2 — Interactive latency budget calculator
def tokens_per_sec_needed(output_tokens: int, max_wait_s: float, ttft_s: float = 0.3) -> float:
    gen_budget = max(max_wait_s - ttft_s, 0.01)
    return output_tokens / gen_budget

for out, wait in [(64, 2.0), (256, 8.0), (1024, 30.0)]:
    print(f"out={out:4} in {wait}s => need ~{tokens_per_sec_needed(out, wait):.1f} tok/s")
````

````python
# Demo 3 — Concurrent users capacity sketch
import math

def max_concurrent(tok_s_system: float, tok_s_per_user: float, headroom: float = 0.7) -> int:
    return max(1, int((tok_s_system * headroom) / tok_s_per_user))

print("users ~", max_concurrent(tok_s_system=2000, tok_s_per_user=40))
````

### Pipeline walkthrough (apply to 01-local-llm-fundamentals)

```
1. Input arrives (user / job / webhook)
2. Normalize + authorize + budget check
3. Gather context (files, RAG, tools, memory)
4. Model / deterministic compute
5. Validate output (schema, policy, tests)
6. Side effects (write, ticket, PR) with authz
7. Observe (metrics, traces, feedback)
8. Learn (eval suite growth, prompt/model revision)
```

**When to compress steps:** tiny internal tools. **When to keep all steps:** multi-tenant or regulated production.


### Local serving advanced notes

**Prefill vs decode** dominate different bottlenecks—size batches for your goal.  
**Quantization** must be validated on *your* tasks (code/math often sensitive).  
**OpenAI-compatible APIs** let you swap Ollama ↔ vLLM ↔ cloud with a router.

| Runtime | Best default use |
|---------|------------------|
| Ollama | Dev laptop UX |
| llama.cpp | Portable/edge |
| vLLM | Multi-user GPU |
| TGI | HF-centric prod features |


````python
# Extra demo — VRAM + context interaction sketch
def kv_cache_gb(layers, heads, dim, seq_len, bytes_per=2, batch=1):
    # very rough educational estimate
    return batch * layers * seq_len * heads * dim * bytes_per * 2 / (1024**3)

print('KV ~', round(kv_cache_gb(32, 32, 128, 8192), 2), 'GB')
````

````python
# Extra demo — OpenAI-compatible client switch
import os

def client_conf(profile: str) -> dict:
    profiles = {
        'ollama': {'base_url': 'http://localhost:11434/v1', 'api_key': 'ollama'},
        'vllm': {'base_url': os.getenv('VLLM_BASE_URL','http://localhost:8000/v1'), 'api_key': os.getenv('VLLM_API_KEY','YOUR_VLLM_API_KEY_HERE')},
        'cloud': {'base_url': 'https://api.openai.com/v1', 'api_key': os.getenv('OPENAI_API_KEY','sk-YOUR_OPENAI_API_KEY_HERE')},
    }
    return profiles[profile]

print(client_conf('ollama'))
````

````python
# Workshop demo — decision scorecard
from dataclasses import dataclass

@dataclass
class Option:
    name: str
    latency: int  # 1=best .. 5=worst
    cost: int
    quality_risk: int
    ops: int
    security: int

def score(o: Option, weights=None) -> float:
    weights = weights or dict(latency=1, cost=1, quality_risk=2, ops=1, security=2)
    return (
        o.latency*weights['latency'] + o.cost*weights['cost'] +
        o.quality_risk*weights['quality_risk'] + o.ops*weights['ops'] +
        o.security*weights['security']
    )

a = Option('A', 2, 3, 2, 2, 2)
b = Option('B', 3, 1, 3, 4, 2)
print(a.name, score(a), b.name, score(b), '-> prefer', a.name if score(a)<score(b) else b.name)
````

### Pitfalls clinic (read aloud)

1. **Metric theater** — optimizing a proxy that users don't feel  
2. **Context stuffing** — more tokens ≠ more truth  
3. **Prompt as security** — never the only control  
4. **Hidden coupling** — tools/models/indexes version-drift  
5. **No rollback** — can't revert prompt/model quickly  
6. **Eval contamination** — testing on training-like snippets  
7. **Happy-path demos** — skipping adversarial & empty-retrieve cases

````python
# Lab helper — MVP scope tracker
from dataclasses import dataclass, field

@dataclass
class MVP:
    must: list[str] = field(default_factory=list)
    should: list[str] = field(default_factory=list)
    defer: list[str] = field(default_factory=list)

    def show(self):
        for label, items in [('MUST', self.must), ('SHOULD', self.should), ('DEFER', self.defer)]:
            print(label)
            for i in items:
                print(' -', i)

mvp = MVP(
    must=['core happy path', 'authn', 'basic eval smoke', 'rollback switch'],
    should=['streaming UX', 'dashboards'],
    defer=['multi-agent', 'perfect personalization', 'every connector'],
)
mvp.show()
````
