## Qwen Family

**Definition.** Alibaba's **Qwen** models include open-weight and service offerings across general, coding, and vision variants—popular in multilingual and open ecosystems.

**Why it matters.** Adds competitive options beyond US-centric vendors, especially for multilingual apps.

**How it works.** Evaluate instruct variants; check licenses; serve via standard inference stacks or cloud APIs.

**Intuition.** A global contender family with broad modality coverage.

**Common pitfalls.**
- Skipping license/region review
- Assuming English-only evals suffice

**When to use.** Multilingual products; open-weight experimentation; regional deployments.

| Line / SKU | Strength focus | Notes |
|------------|----------------|-------|
| Qwen general | Multilingual chat | Verify tier/size |
| Qwen-Coder | Coding | Pair with unit tests |
| Qwen-VL | Vision-language | Multimodal apps |
