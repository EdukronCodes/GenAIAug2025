## Phi Family

**Definition.** **Phi** models are Microsoft's small language models (SLMs) aiming for strong capability at compact sizes.

**Why it matters.** Useful for edge, low-cost, or on-device/near-device scenarios.

**How it works.** Evaluate on your tasks; quantize; consider distillation from larger teachers for specialty skills.

**Intuition.** A smart compact specialist rather than a giant generalist.

**Common pitfalls.**
- Expecting Phi to replace flagship reasoning on all tasks
- Skipping safety evals on SLMs

**When to use.** Local/offline features, cost-sensitive routers, embedded assistants.


````python
# Demo 1 — SLM router
def use_phi(task_complexity: float, private_mode: bool) -> bool:
    return private_mode or task_complexity < 0.4
print(use_phi(0.2, False), use_phi(0.9, True))
````

## Azure OpenAI

**Definition.** **Azure OpenAI** hosts OpenAI models inside Azure with enterprise security, regions, and compliance tooling.

**Why it matters.** Many enterprises standardize here for identity, networking, and contractual controls.

**How it works.** Use Azure endpoints/deployments; authenticate via keys or Entra ID; set content filters.

**Intuition.** Same model brain, corporate castle walls.

**Common pitfalls.**
- Hard-coding keys in notebooks
- Wrong regional deployment for residency

**When to use.** Enterprise production on Azure.

```mermaid
flowchart LR
  App --> AzureOpenAI
  AzureOpenAI --> Deployment
  AzureOpenAI --> Entra
  AzureOpenAI --> Filters
```
