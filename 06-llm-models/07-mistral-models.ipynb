## Notable Models

**Definition.** Mistral AI provides compact high-quality models and larger/mixture variants (names evolve)—often praised for efficiency and developer UX.

**Why it matters.** Attractive when you want strong quality per parameter/dollar.

**How it works.** Use hosted API or self-host open weights; match instruct templates; evaluate MoE serving needs if applicable.

**Intuition.** Sport bikes: fast, efficient, sharp handling.

**Common pitfalls.**
- Assuming MoE is free performance
- Template mismatches causing junk outputs

**When to use.** EU-friendly vendor strategies; efficient self-host; coding/chat apps.

| Line / SKU | Strength focus | Notes |
|------------|----------------|-------|
| Small/medium instruct | Efficient chat | Great defaults |
| Larger / MoE | Higher quality | Serving complexity |
| Codestral-class | Coding | CI evals |
