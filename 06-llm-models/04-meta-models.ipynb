## Llama Family

**Definition.** Meta's Llama models are openly available weights (subject to license) across size tiers, powering self-host and many hosted clones/fine-tunes.

**Why it matters.** Open weights enable private deployment, research, and customization.

**How it works.** Pick parameter size for hardware; apply instruct variants; quantize for GPUs; fine-tune with LoRA.

**Intuition.** A reference chassis the industry builds cars on.

**Common pitfalls.**
- Ignoring license terms
- Running base (non-instruct) models as chatbots

**When to use.** Self-host, air-gapped, or heavy customization needs.

| Line / SKU | Strength focus | Notes |
|------------|----------------|-------|
| Small (7–8B class) | Edge/single GPU | Fast iteration |
| Mid | Stronger quality | More VRAM |
| Large (70B class) | Quality | Multi-GPU / quant |

## Why Llama Matters

**Definition.** Llama catalyzed a vibrant ecosystem: fine-tunes, evals, inference engines, and competitive pressure on APIs.

**Why it matters.** Even if you only use closed APIs, Llama sets baselines and talent practices.

**How it works.** Track ecosystem models; reuse serving stacks; negotiate vendors against open alternatives.

**Intuition.** An open reference point that keeps the market honest.

**Common pitfalls.**
- Equating 'open weights' with 'no obligations'
- Security neglect on self-host

**When to use.** Strategy discussions and platform investments.


```python
# Demo 1 — ecosystem map
eco = ["fine-tunes", "GGUF quants", "vLLM", "eval harnesses", "guards", "data tools"]
print("\n".join(f"- {e}" for e in eco))
```
