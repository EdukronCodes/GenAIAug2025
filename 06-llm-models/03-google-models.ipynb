## Gemini Family

**Definition.** Google's Gemini models emphasize multimodal inputs and scalable context, with multiple size tiers.

**Why it matters.** Strong option when images/audio/video or Google Cloud integration matter.

**How it works.** Select Flash/Pro-class tiers by budget; use multimodal parts; ground with your data tools.

**Intuition.** A multimodal-native product family rather than text-only bolted on.

**Common pitfalls.**
- Sending huge media without compression
- Assuming identical behavior across tiers

**When to use.** Multimodal apps and GCP-centric stacks.

| Line / SKU | Strength focus | Notes |
|------------|----------------|-------|
| Flash-class | Latency/cost | High volume |
| Pro-class | Quality/multimodal | Harder tasks |


````python
# Demo 1 — tier picker
def gemini_tier(need: str) -> str:
    return {"cheap_fast": "flash-class", "quality": "pro-class", "long_multimodal": "pro-class"}.get(need, "flash-class")
print(gemini_tier("cheap_fast"))
````

## Vertex AI

**Definition.** **Vertex AI** is Google Cloud's managed ML/GenAI platform: model access, networking, IAM, and enterprise controls.

**Why it matters.** Enterprises often need VPC-SC, CMEK, and unified cloud billing/governance—not just a public API key.

**How it works.** Deploy/consume models via Vertex endpoints; wire service accounts; set regional constraints.

**Intuition.** The corporate front door to Google models.

**Common pitfalls.**
- Using personal API keys in prod
- Wrong region for data residency

**When to use.** Regulated GCP customers and platform teams.

```mermaid
flowchart LR
  App --> Vertex
  Vertex --> Gemini
  Vertex --> IAM
  Vertex --> Logs
```


````python
# Demo 1 — enterprise checklist
checks = ["region", "IAM least privilege", "CMEK", "logging", "VPC controls", "quotas"]
print("\n".join(f"[ ] {c}" for c in checks))
````
