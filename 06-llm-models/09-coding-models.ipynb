## Notable Coding Models

**Definition.** Coding-specialized or coding-strong models include vendor code SKUs (e.g., Codestral-class, Qwen-Coder-class, DeepSeek-Coder-class, GPT/Claude/Gemini flagships). Names evolve—focus on traits.

**Why it matters.** Code assistants need different strengths than general chat.

**How it works.** Bake off on your languages, repos, and tool workflows (edit, test, repair).

**Intuition.** A mechanic's toolset, not a novelist quill.

**Common pitfalls.**
- Judging only on toy HumanEval snippets
- No repository context in evals

**When to use.** IDE copilots, code review bots, migration agents.

| Line / SKU | Strength focus | Notes |
|------------|----------------|-------|
| General flagship | Broad reasoning | Costly |
| Code-specialized | Completion/FIM | Repo evals needed |
| Open coder lines | Customizable | Self-host ops |

## Eval Signals

**Definition.** Useful signals: pass@k on unit tests, repo repair rate, build green rate, security regressions, latency to first patch, human accept rate.

**Why it matters.** Leaderboards rarely match your frameworks and private code conventions.

**How it works.** Build a harness that runs generated patches against tests in sandboxes.

**Intuition.** Green tests beat vibes.

**Common pitfalls.**
- Optimizing only for short algorithmic puzzles
- No sandbox → unsafe eval

**When to use.** Before enabling any coding agent in CI.

| Signal | Captures |
|--------|----------|
| pass@k | Functional correctness |
| build green | Integration reality |
| accept rate | Dev UX |
| vuln patterns | Safety |
| latency | Interactive fit |


```python
# Demo 1 — pass@k estimator
import random
def pass_at_k(n_ok, n_total, k):
    # simple ratio of tasks with >=1 success if you only store best-of — educational
    return n_ok / n_total
print("pass@1", pass_at_k(12, 20, 1))
```

```python
# Demo 2 — repo eval record
rec = {"task_id": "fix-null-check", "tests_passed": True, "files_touched": 2, "latency_s": 38}
print(rec)
```

```python
# Demo 3 — security gate
forbidden = ["os.system(", "eval(", "subprocess.Popen(shell=True"]
patch = "x = eval(user)"  # bad
print("blocked", any(f in patch for f in forbidden))
```

```python
# Demo 4 — accept rate
shown, accepted = 80, 35
print("accept_rate", accepted/shown)
```
