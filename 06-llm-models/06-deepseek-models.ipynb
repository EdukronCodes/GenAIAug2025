## DeepSeek Lines

**Definition.** DeepSeek releases competitive LLM lines (chat, reasoning, coding-focused variants) that are influential in open and API markets—confirm current names and licenses.

**Why it matters.** Offers strong quality/cost contenders for coding and reasoning workloads.

**How it works.** Bake off on your harness; watch tokenizer/template differences; monitor safety behaviors.

**Intuition.** A high-performance challenger brand in the open ecosystem.

**Common pitfalls.**
- Blind swaps without regression tests
- Ignoring data handling terms

**When to use.** Cost-sensitive coding/reasoning features with solid eval coverage.

| Line / SKU | Strength focus | Notes |
|------------|----------------|-------|
| Chat-class | General assistant | Default |
| Reasoner-class | Hard multi-step | Latency/cost trade |
| Coder-class | Repositories/tests | Pair with CI |

```python
# Demo 4 — regression gate before swap
baseline = {"pass@1": 0.61, "faithfulness": 0.74}
candidate = {"pass@1": 0.66, "faithfulness": 0.72}
def allow_swap(b, c, min_faith=0.73):
    return c["pass@1"] >= b["pass@1"] and c["faithfulness"] >= min_faith
print("swap?", allow_swap(baseline, candidate))
```
