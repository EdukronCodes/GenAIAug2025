## Claude Family

**Definition.** Anthropic's Claude models are offered in size/speed tiers (Haiku/Sonnet/Opus-class naming historically) for chat, coding, and long-context work—verify current IDs.

**Why it matters.** A major hosted alternative with strong instruction-following and coding reputation.

**How it works.** Pick tier by latency/cost/quality; use Messages API; enable tools as needed.

**Intuition.** Good/better/best lanes tuned for thoughtful assistants.

**Common pitfalls.**
- Assuming all tiers share identical context/tool behavior
- Skipping prompt caching opportunities

**When to use.** Assistants needing careful writing, coding, or long documents.

| Line / SKU | Strength focus | Notes |
|------------|----------------|-------|
| Haiku-class | Speed / cost | High QPS |
| Sonnet-class | Balance | Default many apps |
| Opus-class | Hard tasks | Selective use |


```python
# Demo 1 — tier router
def claude_tier(sla: str) -> str:
    return {"realtime": "haiku-class", "balanced": "sonnet-class", "max_quality": "opus-class"}[sla]
print(claude_tier("balanced"))
```

## Distinguishing Traits

**Definition.** Practically, teams often cite constitutional/safety emphasis, strong writing/coding UX, tool use, and long-context document workflows as Claude differentiators—always validate on your evals.

**Why it matters.** Differentiation informs vendor shortlists and dual-vendor strategies.

**How it works.** Benchmark on *your* tasks: groundedness, coding tests, refusal quality, latency.

**Intuition.** Taste + safety posture + engineering affordances—not just leaderboard lore.

**Common pitfalls.**
- Choosing vendors from Twitter anecdotes alone
- Ignoring data residency requirements

**When to use.** When writing quality, careful refusals, or long docs dominate requirements.


```python
# Demo 1 — eval dimensions for vendor compare
dims = ["faithfulness", "coding_pass@1", "refusal_quality", "latency_p95", "cost_per_1k"]
print(dims)
```

```python
# Demo 2 — dual vendor router
def vendor_route(doc_pages: int, need_tools: bool) -> str:
    if doc_pages > 100:
        return "prefer_long_context_family"
    if need_tools:
        return "either_with_tool_evals"
    return "cheapest_that_passes_eval"
print(vendor_route(120, False), vendor_route(3, True))
```
