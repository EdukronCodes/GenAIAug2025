## 1. Cost & Latency Levers

| Lever | Effect |
|-------|--------|
| Exact cache | Free repeats |
| Semantic cache | Near-duplicate FAQs |
| Smaller models | Lower $ / higher speed |
| Shorter prompts | Less prefill |
| Batch | Higher throughput |
| Speculative decode | Faster tokens |


````python
# Demo 1 — Semantic cache (embedding cosine) toy
import numpy as np
from numpy.linalg import norm

def cos(a, b):
    return float(np.dot(a, b) / (norm(a) * norm(b) + 1e-9))

class SemanticCache:
    def __init__(self, thresh=0.9):
        self.thresh = thresh
        self.items: list[tuple[np.ndarray, str]] = []

    def get(self, emb, default=None):
        for e, val in self.items:
            if cos(emb, e) >= self.thresh:
                return val
        return default

    def put(self, emb, val):
        self.items.append((emb, val))

c = SemanticCache()
c.put(np.array([1.0, 0.0]), "cached-answer")
print(c.get(np.array([0.99, 0.01])), c.get(np.array([0.0, 1.0])))
````

## 2. Model Cascades

Route easy traffic to small/cheap models; escalate hard cases.


````python
# Demo 2 — Cascade router
def cascade_route(intent_confidence: float, needs_tools: bool) -> str:
    if needs_tools:
        return "tool_capable_large"
    if intent_confidence > 0.85:
        return "small_local"
    return "medium_cloud"

print(cascade_route(0.9, False), cascade_route(0.4, False), cascade_route(0.9, True))
````

## 3. Reliability Knobs

Retries with jitter, hedged requests, graceful degradation (retrieve-only), fallback templates, circuit breakers.


````python
# Demo 3 — Circuit breaker
from dataclasses import dataclass
import time

@dataclass
class Breaker:
    fail_threshold: int = 3
    cool_s: float = 5
    failures: int = 0
    open_until: float = 0.0

    def allow(self) -> bool:
        return time.time() >= self.open_until

    def fail(self):
        self.failures += 1
        if self.failures >= self.fail_threshold:
            self.open_until = time.time() + self.cool_s
            self.failures = 0

b = Breaker()
for _ in range(3):
    b.fail()
print("allow after open", b.allow(), "open_until", b.open_until)
````

````python
# Workshop demo — decision scorecard
from dataclasses import dataclass

@dataclass
class Option:
    name: str
    latency: int  # 1=best .. 5=worst
    cost: int
    quality_risk: int
    ops: int
    security: int

def score(o: Option, weights=None) -> float:
    weights = weights or dict(latency=1, cost=1, quality_risk=2, ops=1, security=2)
    return (
        o.latency*weights['latency'] + o.cost*weights['cost'] +
        o.quality_risk*weights['quality_risk'] + o.ops*weights['ops'] +
        o.security*weights['security']
    )

a = Option('A', 2, 3, 2, 2, 2)
b = Option('B', 3, 1, 3, 4, 2)
print(a.name, score(a), b.name, score(b), '-> prefer', a.name if score(a)<score(b) else b.name)
````

### Pitfalls clinic (read aloud)

1. **Metric theater** — optimizing a proxy that users don't feel  
2. **Context stuffing** — more tokens ≠ more truth  
3. **Prompt as security** — never the only control  
4. **Hidden coupling** — tools/models/indexes version-drift  
5. **No rollback** — can't revert prompt/model quickly  
6. **Eval contamination** — testing on training-like snippets  
7. **Happy-path demos** — skipping adversarial & empty-retrieve cases

````python
# Lab helper — MVP scope tracker
from dataclasses import dataclass, field

@dataclass
class MVP:
    must: list[str] = field(default_factory=list)
    should: list[str] = field(default_factory=list)
    defer: list[str] = field(default_factory=list)

    def show(self):
        for label, items in [('MUST', self.must), ('SHOULD', self.should), ('DEFER', self.defer)]:
            print(label)
            for i in items:
                print(' -', i)

mvp = MVP(
    must=['core happy path', 'authn', 'basic eval smoke', 'rollback switch'],
    should=['streaming UX', 'dashboards'],
    defer=['multi-agent', 'perfect personalization', 'every connector'],
)
mvp.show()
````
