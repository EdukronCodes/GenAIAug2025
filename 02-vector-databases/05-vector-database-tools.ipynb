## FAISS

**Definition.** High-performance ANN *library*, not a DB server.

**Why it matters.** Speed/control.

**How it works.** Index.add/search; you own metadata/ops.

**Intuition.** Engine without the car.

**Common pitfalls.**
- Expecting filters

**When to use.** Custom/GPU/offline.

## Pinecone

**Definition.** Managed vector DB service.

**Why it matters.** Ops-light scale.

**How it works.** Upsert/query + namespaces/filters.

**Intuition.** Cloud utility.

**Common pitfalls.**
- Cost surprises

**When to use.** Managed teams.


```python
import json
print('Api-Key: YOUR_API_KEY')
print(json.dumps({'vectors':[{'id':'a','values':[0.1,0.2],'metadata':{'lang':'en'}}],'namespace':'prod'},indent=2))
```

## Qdrant

**Definition.** Open-source vector DB with strong payloads/filters.

**Why it matters.** Prod RAG filters.

**How it works.** Collections + points + filter query.

**Intuition.** Real DB for vectors.

**Common pitfalls.**
- Missing payload indexes

**When to use.** Self-host/cloud + filters.


```python
import json
print(json.dumps({'id':1,'vector':[0.1,0.2],'payload':{'city':'Berlin'}},indent=2))
```

## Weaviate

**Definition.** Schema/classes + hybrid search vector DB.

**Why it matters.** Hybrid products.

**How it works.** Schema → objects → nearVector/hybrid.

**Intuition.** Search DNA + vectors.

**Common pitfalls.**
- Module vectorizer drift

**When to use.** Schema-rich hybrid apps.


```python
import json
print(json.dumps({'query':'refund','alpha':0.5,'class':'Article'},indent=2))
```
