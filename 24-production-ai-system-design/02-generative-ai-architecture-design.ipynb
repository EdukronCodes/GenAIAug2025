## 2. Model Gateway Responsibilities

Auth to providers, routing/cascades, timeouts, retries, budgeting, prompt templates registry, abuse policy, shadow traffic.


````python
# Demo 1 — Gateway interface sketch
from dataclasses import dataclass

@dataclass
class ChatRequest:
    tenant: str
    messages: list[dict]
    model_hint: str | None = None

@dataclass
class ChatResponse:
    content: str
    model: str
    tokens_in: int
    tokens_out: int

class ModelGateway:
    def __init__(self, routes: dict[str, str]):
        self.routes = routes

    def select(self, req: ChatRequest) -> str:
        return req.model_hint or self.routes.get(req.tenant, "default")

    def complete(self, req: ChatRequest) -> ChatResponse:
        model = self.select(req)
        return ChatResponse("(stub)", model, 10, 5)

gw = ModelGateway({"acme": "cloud-large", "startups": "local-small"})
print(gw.complete(ChatRequest("acme", [{"role": "user", "content": "hi"}])))
````

## 4. Failure Design

Provider outages, quality regressions, retrieval empty, tool timeouts. Patterns: fallbacks, cached answers, degraded modes, circuit breakers.


````python
# Demo 2 — Degraded mode selector
def mode(provider_ok: bool, index_ok: bool) -> str:
    if provider_ok and index_ok:
        return "full_rag"
    if provider_ok and not index_ok:
        return "llm_only_with_warning"
    if not provider_ok and index_ok:
        return "retrieve_only_snippets"
    return "maintenance_banner"

print(mode(True, False), mode(False, True), mode(False, False))
````

### Pipeline walkthrough (apply to 02-generative-ai-architecture-design)

```
1. Input arrives (user / job / webhook)
2. Normalize + authorize + budget check
3. Gather context (files, RAG, tools, memory)
4. Model / deterministic compute
5. Validate output (schema, policy, tests)
6. Side effects (write, ticket, PR) with authz
7. Observe (metrics, traces, feedback)
8. Learn (eval suite growth, prompt/model revision)
```

**When to compress steps:** tiny internal tools. **When to keep all steps:** multi-tenant or regulated production.


### System design advanced notes

Always start from **job-to-be-done + constraints + workload numbers**.  
Produce an MVP path and a scale path.  
Close with **evals, security, rollout**.

**10-box vocabulary:** client, gateway, orchestrator, retriever, ranker, model gateway, tool bus, memory, obs, eval worker.


````python
# Extra demo — interview math kit
def tokens_per_day(qps, in_tok, out_tok):
    return qps*86400*(in_tok+out_tok)

def gpus_needed(tok_per_day, tok_per_s_per_gpu, utilization=0.6):
    need = tok_per_day / 86400
    return max(1, round(need / (tok_per_s_per_gpu * utilization)))

tpd = tokens_per_day(5, 2000, 500)
print('tokens/day', tpd, 'gpus~', gpus_needed(tpd, 2000))
````

````python
# Workshop demo — decision scorecard
from dataclasses import dataclass

@dataclass
class Option:
    name: str
    latency: int  # 1=best .. 5=worst
    cost: int
    quality_risk: int
    ops: int
    security: int

def score(o: Option, weights=None) -> float:
    weights = weights or dict(latency=1, cost=1, quality_risk=2, ops=1, security=2)
    return (
        o.latency*weights['latency'] + o.cost*weights['cost'] +
        o.quality_risk*weights['quality_risk'] + o.ops*weights['ops'] +
        o.security*weights['security']
    )

a = Option('A', 2, 3, 2, 2, 2)
b = Option('B', 3, 1, 3, 4, 2)
print(a.name, score(a), b.name, score(b), '-> prefer', a.name if score(a)<score(b) else b.name)
````

### Pitfalls clinic (read aloud)

1. **Metric theater** — optimizing a proxy that users don't feel  
2. **Context stuffing** — more tokens ≠ more truth  
3. **Prompt as security** — never the only control  
4. **Hidden coupling** — tools/models/indexes version-drift  
5. **No rollback** — can't revert prompt/model quickly  
6. **Eval contamination** — testing on training-like snippets  
7. **Happy-path demos** — skipping adversarial & empty-retrieve cases

### Glossary boost

| Term | Expanded meaning |
|------|------------------|
| Canary | Partial traffic to a new variant with automatic rollback |
| Golden set | Versioned labeled examples for regression |

| Shadow traffic | New system sees traffic but doesn't affect users |
| Circuit breaker | Stop calling a failing dependency temporarily |

````python
# Lab helper — MVP scope tracker
from dataclasses import dataclass, field

@dataclass
class MVP:
    must: list[str] = field(default_factory=list)
    should: list[str] = field(default_factory=list)
    defer: list[str] = field(default_factory=list)

    def show(self):
        for label, items in [('MUST', self.must), ('SHOULD', self.should), ('DEFER', self.defer)]:
            print(label)
            for i in items:
                print(' -', i)

mvp = MVP(
    must=['core happy path', 'authn', 'basic eval smoke', 'rollback switch'],
    should=['streaming UX', 'dashboards'],
    defer=['multi-agent', 'perfect personalization', 'every connector'],
)
mvp.show()
````
