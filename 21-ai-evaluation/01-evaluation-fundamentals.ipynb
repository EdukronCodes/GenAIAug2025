# Evaluation Fundamentals

Evaluating generative AI is harder than classic ML metrics: outputs are open-ended, judges disagree, and production traffic drifts. This notebook establishes types of eval, dataset hygiene, and release-gate philosophy.



## Learning Objectives

- Explain why GenAI evaluation is hard

## 1. Why Evaluation Is Hard for GenAI

| Challenge | Implication |
|-----------|-------------|
| Open-ended outputs | Exact match fails |
| Multiple valid answers | Need rubrics / semantic metrics |
| Tool-using agents | Must score trajectories |
| Stochastic decoding | Flaky CI without seeds/budgets |
| Domain shift | Public benches ≠ your users |
| Safety vs helpfulness | Tradeoffs require policy |

### Intuition
Treat eval like **product QA + scientific experiment**: fixed fixtures, versioned datasets, confidence intervals, and human adjudication for edge cases.


## 2. Evaluation Types

| Type | Definition | When |
|------|------------|------|
| Unit / deterministic | Schema, regex, parsers | Always |
| Golden set offline | Curated Q→rubric | Pre-release |
| Model-as-judge | LLM scores outputs | Scale review |
| Human eval | Expert ratings | High-risk |
| Online | Thumbs, CSAT, task success | Production |
| Adversarial | Red team / jailbreaks | Security |
| A/B | Compare variants | Product |


````python
# Demo 1 — Minimal evaluation harness structure
from dataclasses import dataclass, field
from typing import Any, Callable

@dataclass
class Example:
    id: str
    input: dict[str, Any]
    gold: dict[str, Any] | None = None

@dataclass
class EvalResult:
    example_id: str
    score: float
    pass_: bool
    info: dict[str, Any] = field(default_factory=dict)

Metric = Callable[[Example, Any], EvalResult]

def run_eval(examples: list[Example], predict: Callable[[Example], Any], metric: Metric) -> dict:
    results = [metric(ex, predict(ex)) for ex in examples]
    rate = sum(r.pass_ for r in results) / max(len(results), 1)
    return {"n": len(results), "pass_rate": rate, "results": results}

examples = [Example("1", {"q": "2+2"}, {"a": "4"}), Example("2", {"q": "3+1"}, {"a": "4"})]
predict = lambda ex: {"a": "4"}
metric = lambda ex, pred: EvalResult(ex.id, 1.0, pred.get("a") == (ex.gold or {}).get("a"))
print(run_eval(examples, predict, metric)["pass_rate"])
````

## 3. Dataset Hygiene

1. **Version** datasets (`evals/v3/…`) and pin in CI
2. **Deduplicate** near-duplicates that leak train/test
3. **Balance** difficulty, topics, languages
4. **Label guide** — written rubric with examples
5. **PII scrub** before committing
6. **Separate** unit fixtures vs expensive judge sets

## 4. Gate Philosophy

- **Hard gates:** security, PII leaks, schema validity
- **Soft gates:** judge score ≥ threshold with CI bootstrap
- **Canary:** online metrics before full rollout
- Never gate solely on a single LLM-judge without spot checks


````python
# Demo 3 — Gate combiner
def gate(security_ok: bool, schema_ok: bool, judge_score: float, min_judge: float = 0.8) -> str:
    if not security_ok or not schema_ok:
        return "BLOCK"
    if judge_score < min_judge:
        return "BLOCK_SOFT"
    return "ALLOW"

print(gate(True, True, 0.85), gate(True, False, 0.99), gate(True, True, 0.5))
````

### Pipeline walkthrough (apply to 01-evaluation-fundamentals)

```
1. Input arrives (user / job / webhook)
2. Normalize + authorize + budget check
3. Gather context (files, RAG, tools, memory)
4. Model / deterministic compute
5. Validate output (schema, policy, tests)
6. Side effects (write, ticket, PR) with authz
7. Observe (metrics, traces, feedback)
8. Learn (eval suite growth, prompt/model revision)
```

**When to compress steps:** tiny internal tools. **When to keep all steps:** multi-tenant or regulated production.

````python
# Extra demo — bootstrap mean for flaky judge scores
import random

def bootstrap_mean(scores, n=500, rng=random.Random(0)):
    means = []
    for _ in range(n):
        sample = [scores[rng.randrange(len(scores))] for _ in scores]
        means.append(sum(sample)/len(sample))
    means.sort()
    return means[int(0.025*n)], sum(means)/len(means), means[int(0.975*n)]

print(bootstrap_mean([0.7,0.8,0.75,0.9,0.6,0.85]))
````

````python
# Extra demo — confusion-style agreement human vs judge
def agreement(human, judge):
    assert len(human)==len(judge)
    return sum(h==j for h,j in zip(human,judge))/len(human)

print(agreement([1,1,0,1],[1,0,0,1]))
````

````python
# Workshop demo — decision scorecard
from dataclasses import dataclass

@dataclass
class Option:
    name: str
    latency: int  # 1=best .. 5=worst
    cost: int
    quality_risk: int
    ops: int
    security: int

def score(o: Option, weights=None) -> float:
    weights = weights or dict(latency=1, cost=1, quality_risk=2, ops=1, security=2)
    return (
        o.latency*weights['latency'] + o.cost*weights['cost'] +
        o.quality_risk*weights['quality_risk'] + o.ops*weights['ops'] +
        o.security*weights['security']
    )

a = Option('A', 2, 3, 2, 2, 2)
b = Option('B', 3, 1, 3, 4, 2)
print(a.name, score(a), b.name, score(b), '-> prefer', a.name if score(a)<score(b) else b.name)
````

### Pitfalls clinic (read aloud)

1. **Metric theater** — optimizing a proxy that users don't feel  
2. **Context stuffing** — more tokens ≠ more truth  
3. **Prompt as security** — never the only control  
4. **Hidden coupling** — tools/models/indexes version-drift  
5. **No rollback** — can't revert prompt/model quickly  
6. **Eval contamination** — testing on training-like snippets  
7. **Happy-path demos** — skipping adversarial & empty-retrieve cases

````python
# Lab helper — MVP scope tracker
from dataclasses import dataclass, field

@dataclass
class MVP:
    must: list[str] = field(default_factory=list)
    should: list[str] = field(default_factory=list)
    defer: list[str] = field(default_factory=list)

    def show(self):
        for label, items in [('MUST', self.must), ('SHOULD', self.should), ('DEFER', self.defer)]:
            print(label)
            for i in items:
                print(' -', i)

mvp = MVP(
    must=['core happy path', 'authn', 'basic eval smoke', 'rollback switch'],
    should=['streaming UX', 'dashboards'],
    defer=['multi-agent', 'perfect personalization', 'every connector'],
)
mvp.show()
````
