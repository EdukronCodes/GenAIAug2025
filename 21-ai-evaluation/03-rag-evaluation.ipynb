## 1. Retrieval Metrics

| Metric | Question answered |
|--------|-------------------|
| Precision@K | Of top K, how many relevant? |
| Recall@K | Of all relevant, how many in top K? |
| Hit rate | Any relevant in top K? |
| MRR | How early is first relevant? |
| nDCG@K | Graded relevance quality |


````python
# Demo 1 — Retrieval metrics
import math
from typing import Sequence

def precision_at_k(ranked: Sequence[str], relevant: set[str], k: int) -> float:
    top = ranked[:k]
    return sum(1 for x in top if x in relevant) / max(k, 1)

def recall_at_k(ranked: Sequence[str], relevant: set[str], k: int) -> float:
    if not relevant:
        return 0.0
    return sum(1 for x in ranked[:k] if x in relevant) / len(relevant)

def hit_at_k(ranked: Sequence[str], relevant: set[str], k: int) -> float:
    return float(any(x in relevant for x in ranked[:k]))

def mrr(ranked: Sequence[str], relevant: set[str]) -> float:
    for i, x in enumerate(ranked, 1):
        if x in relevant:
            return 1.0 / i
    return 0.0

def dcg(rels: list[float]) -> float:
    return sum((2**r - 1) / math.log2(i+1) for i, r in enumerate(rels, 1))

def ndcg_at_k(ranked: Sequence[str], graded: dict[str, float], k: int) -> float:
    rels = [graded.get(x, 0.0) for x in ranked[:k]]
    ideal = sorted(graded.values(), reverse=True)[:k]
    denom = dcg(ideal)
    return 0.0 if denom == 0 else dcg(rels) / denom

ranked = ["d1", "d2", "d3", "d4"]
rel = {"d2", "d4"}
graded = {"d2": 3.0, "d4": 1.0, "d1": 0.0}
print(precision_at_k(ranked, rel, 3), recall_at_k(ranked, rel, 3), hit_at_k(ranked, rel, 2), mrr(ranked, rel), ndcg_at_k(ranked, graded, 3))
````

## 2. Context Metrics

Context precision/recall ask whether retrieved *evidence sentences* actually support the answer—not just whether the doc id was right.


````python
# Demo 2 — Context precision/recall with gold evidence sentences
def context_precision(retrieved_sents: list[str], gold_sents: set[str]) -> float:
    if not retrieved_sents:
        return 0.0
    return sum(1 for s in retrieved_sents if s in gold_sents) / len(retrieved_sents)

def context_recall(retrieved_sents: list[str], gold_sents: set[str]) -> float:
    if not gold_sents:
        return 0.0
    return sum(1 for s in gold_sents if s in retrieved_sents) / len(gold_sents)

ret = ["Eiffel Tower is in Paris", "It is tall"]
gold = {"Eiffel Tower is in Paris", "Built in 1889"}
print(context_precision(ret, gold), context_recall(ret, gold))
````

## 3. Generation Metrics for RAG

- **Faithfulness** — grounded in context
- **Answer relevance** — addresses question
- **Citation quality** — correct pointers


````python
# Demo 3 — Tiny faithfulness heuristic for RAG answers
import re

def faithfulness(answer: str, context: str) -> float:
    sents = [s.strip() for s in re.split(r"[.!?]", answer) if s.strip()]
    ctx = set(context.lower().split())
    if not sents:
        return 0.0
    ok = sum(1 for s in sents if len(set(s.lower().split()) & ctx) >= 2)
    return ok / len(sents)

print(faithfulness("Paris is in France. Mars is wet.", "Paris is a city in France."))
````

## 4. End-to-End Scorecard

| Stage | Metric | Target example |
|-------|--------|----------------|
| Retrieval | hit@5 | ≥0.90 |
| Context | ctx recall | ≥0.80 |
| Generation | faithfulness | ≥0.85 |
| Product | user thumbs up | ≥0.70 |


````python
# Demo 4 — Scorecard aggregator
def scorecard(rows: list[dict]) -> dict:
    keys = rows[0].keys()
    return {k: sum(r[k] for r in rows)/len(rows) for k in keys}

print(scorecard([
    {"hit@5": 1, "faith": 0.8},
    {"hit@5": 0, "faith": 0.9},
]))
````

````python
# Extra demo — bootstrap mean for flaky judge scores
import random

def bootstrap_mean(scores, n=500, rng=random.Random(0)):
    means = []
    for _ in range(n):
        sample = [scores[rng.randrange(len(scores))] for _ in scores]
        means.append(sum(sample)/len(sample))
    means.sort()
    return means[int(0.025*n)], sum(means)/len(means), means[int(0.975*n)]

print(bootstrap_mean([0.7,0.8,0.75,0.9,0.6,0.85]))
````

````python
# Extra demo — confusion-style agreement human vs judge
def agreement(human, judge):
    assert len(human)==len(judge)
    return sum(h==j for h,j in zip(human,judge))/len(human)

print(agreement([1,1,0,1],[1,0,0,1]))
````

````python
# Workshop demo — decision scorecard
from dataclasses import dataclass

@dataclass
class Option:
    name: str
    latency: int  # 1=best .. 5=worst
    cost: int
    quality_risk: int
    ops: int
    security: int

def score(o: Option, weights=None) -> float:
    weights = weights or dict(latency=1, cost=1, quality_risk=2, ops=1, security=2)
    return (
        o.latency*weights['latency'] + o.cost*weights['cost'] +
        o.quality_risk*weights['quality_risk'] + o.ops*weights['ops'] +
        o.security*weights['security']
    )

a = Option('A', 2, 3, 2, 2, 2)
b = Option('B', 3, 1, 3, 4, 2)
print(a.name, score(a), b.name, score(b), '-> prefer', a.name if score(a)<score(b) else b.name)
````

### Pitfalls clinic (read aloud)

1. **Metric theater** — optimizing a proxy that users don't feel  
2. **Context stuffing** — more tokens ≠ more truth  
3. **Prompt as security** — never the only control  
4. **Hidden coupling** — tools/models/indexes version-drift  
5. **No rollback** — can't revert prompt/model quickly  
6. **Eval contamination** — testing on training-like snippets  
7. **Happy-path demos** — skipping adversarial & empty-retrieve cases

````python
# Lab helper — MVP scope tracker
from dataclasses import dataclass, field

@dataclass
class MVP:
    must: list[str] = field(default_factory=list)
    should: list[str] = field(default_factory=list)
    defer: list[str] = field(default_factory=list)

    def show(self):
        for label, items in [('MUST', self.must), ('SHOULD', self.should), ('DEFER', self.defer)]:
            print(label)
            for i in items:
                print(' -', i)

mvp = MVP(
    must=['core happy path', 'authn', 'basic eval smoke', 'rollback switch'],
    should=['streaming UX', 'dashboards'],
    defer=['multi-agent', 'perfect personalization', 'every connector'],
)
mvp.show()
````
